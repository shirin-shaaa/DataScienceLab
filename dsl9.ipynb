```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

# Split the data
msg_train, msg_test, label_train, label_test = train_test_split(
    df['message'],
    df['label'],
    test_size=0.2,
    random_state=42
)
# Create pipeline
pipeline = Pipeline([
    ('bow', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('classifier', MultinomialNB())
])

# Train the model
pipeline.fit(msg_train, label_train)

# Predict test data
predictions = pipeline.predict(msg_test)

# Classification report
print(classification_report(label_test, predictions))
```

```python
from sklearn import model_selection, naive_bayes, svm
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.metrics import accuracy_score, classification_report

# Create SVM pipeline
pipeline1 = Pipeline([
    ('bow', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('classifier', svm.SVC(C=1.0, kernel='linear'))
])

# Train the model
pipeline1.fit(msg_train, label_train)

# Predict test data
predictions1 = pipeline1.predict(msg_test)

# Accuracy
print("Accuracy:", accuracy_score(label_test, predictions1))

# Classification report
print(classification_report(label_test, predictions1))
```
