```python


x = np.arange(0, 10, 1)

plt.plot(
    x, x+3,
    color='b',
    lw=3,
    linestyle='dashed',   # Use either linestyle or ls, not both
    marker='o',
    markersize=12,
    markeredgecolor='green',
    markeredgewidth=4,
    markerfacecolor='red'
)

plt.plot(
    x, x+4,
    color='r',
    lw=4,
    linestyle='dashed',
    marker='o',
    markersize=10,
    markeredgecolor='blue',
    markeredgewidth=5,
    markerfacecolor='green'
)

plt.show()
```

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.displot(tips["tip"])
```

```python
sns.distplot(tips['total_bill'],kde=False,bins=30)
```

```python
sns.histplot(tips['total_bill'], bins=30, kde=False)
plt.show()
```

```python
sns.jointplot(x ='total_bill',y='tip',data=tips,kind='scatter')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='hex')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='reg')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='resid')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='kde')
```

```python

sns.pairplot(tips)
```

```python
sns.rugplot(tips['total_bill'])
```

```python
sns.barplot(x= 'sex',y='total_bill',data=tips,
    hue='sex',
    palette='Set2',
    legend=False)
```

```python
import numpy as np
sns.barplot(x= 'sex',y='total_bill',data=tips,estimator=np.std,
    hue='sex',
    palette='Set2',
    legend=False)
```

```python
sns.countplot(x='sex',data=tips,
    hue='sex',
    palette='Set2',
    legend=False)
```

```python
sns.boxplot(x='day',y='total_bill',data=tips,palette='rainbow')
```

```python
sns.boxplot(data=tips,palette='rainbow',orient='h')
```

```python
sns.boxplot(x='day',y='total_bill',hue ='smoker',
data=tips,palette='coolwarm')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,palette='rainbow')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,hue =
'sex',palette='Set1')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,hue = 'sex',split=
True,palette='Set1')
```

```python
sns.stripplot(
    x='day',
    y='total_bill',
    data=tips,
    hue='day',
    palette='Set1',
    legend=False
)

plt.show()
```

```python
sns.stripplot(
    x='day',
    y='total_bill',
    data=tips,
    jitter=True,
    hue='day',
    palette='Set1',
    legend=False
)

plt.show()
```

```python
sns.stripplot(x='day',
y='total_bill',data=tips,jitter=True,hue='sex',palette='Set1')
```

```python
sns.stripplot(
    x='day',
    y='total_bill',
    data=tips,
    hue='sex',
    jitter=True,
    dodge=True,
    palette='Set1'
)

plt.show()
```

```python
sns.catplot(
    x='sex',
    y='total_bill',
    data=tips,
    kind='bar',
    hue='sex',
    palette=['red', 'green'],
    legend=False
)

plt.show()
```
